# GIF Picture Creator for Showcasing Simulation Steps

```python
# Setup
from schema import *
```

```python
# Simulation parameters:
# Width and height of the environment
# Number of survivors, safe zones and robots (agents)
environment = EnvironmentModel(
    width=8,
    height=8,
    n_survivors=4,
    n_safe_zones=2, # must be >1 for task 4
    n_robot_agents=1, # must be 1 for task 4
    seed=42,
)

environment.visualize_graph("output/gif_images/graph_visualization_step_0.png")
```

```python
i: int = 0

while not environment.all_survivors_rescued():
    # do one step
    i += 1
    environment.step()
    environment.visualize_graph(f"output/gif_images/graph_visualization_step_{i}.png")
```
